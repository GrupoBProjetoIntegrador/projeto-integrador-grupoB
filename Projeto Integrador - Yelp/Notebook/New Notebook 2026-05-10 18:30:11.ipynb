```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
import numpy as np

# --- INPUT CONFIG ---
# Escolha os schemas e tabelas de input para o diagrama
input_config = {
    'bronze': {
        'schema': 'workspace.yelp_bronze',
        'tables': [
            {
                'name': 'yelp_academic_dataset_business',
                'columns': ['🔑 business_id', 'address', 'attributes', 'categories', 'city', 'hours', 'is_open', 'latitude', 'longitude', 'name', 'postal_code', 'review_count', 'stars', 'state', 'hora_ingestao']
            },
            {
                'name': 'yelp_academic_dataset_review',
                'columns': ['🔑 review_id', '🔗 business_id', '🔗user_id', 'cool', 'date',  'funny', 'stars', 'text', 'useful', 'hora_ingestao']
            },
            {
                'name': 'yelp_academic_dataset_user',
                'columns': ['🔑 user_id', 'average_stars', 'compliment_cool', 'compliment_cute', 'compliment_funny', 'compliment_hot', 'compliment_list', 'compliment_more', 'compliment_note', 'compliment_photos', 'compliment_plain', 'compliment_profile', 'compliment_writer', 'cool', 'elite', 'fans', 'friends', 'funny', 'name', 'review_count', 'useful', 'yelping_since', 'hora_ingestao']
            }
        ]
    },
    'silver': {
        'schema': 'workspace.yelp_silver',
        'tables': [
            {
                'name': 'business',
                'columns': ['🔑 business_id', 'hours', 'review_count', 'stars', 'state', 'hora_ingestao', 'food_category', 'name', 'city', 'data_processamento_silver']
            },
            {
                'name': 'review',
                'columns': ['🔑 review_id', '🔗 business_id', 'cool', 'date', 'funny', 'stars',  'useful', '🔗 user_id', 'hora_ingestao','food_category','data_processamento_silver']
            },
            {
                'name': 'user',
                'columns': ['🔑 user_id', 'average_stars', 'elite', 'fans', 'name', 'review_count', 'yelping_since', 'hora_ingestao', 'review_food_count', 'data_processamento_silver']
            }
        ]
    },
    'gold': {
        'schema': 'workspace.yelp_gold',
        'tables': [
            {
                'name': 'gold_business_reviews_integrado',
                'columns': ['🔑 review_id', 'estabelecimento', 'food_category', 'stars', 
                            'date', 'dia_semana', 'dia_semana_numero', 'estado', 'data_processamento_silver', 'macrorregiao']
            },
            {
                'name': 'marketing',
                'columns': ['🔑 business_id', 'name', 'food_category', 'city', 'state',
                            'avg_stars', 'review_count', 'sum_stars', 'rating_yelp']
            },
            {
                'name': 'user_behavior_complete',
                'columns': ['🔑 user_id', 'name', 'state', 'city', 'scope', 'score', 'review_food_count', 'review_cool', 'review_funny', 'review_useful', 'fans', 'average_stars', 'yeslping_since', 'elite', 'is_top_20_percent']
            }
        ]
    }
}

# Create figure
fig, ax = plt.subplots(figsize=(18, 12))
ax.set_xlim(0, 18)
ax.set_ylim(0, 12)
ax.axis('off')

# Color scheme
bronze_color = '#CD7F32'
silver_color = '#C0C0C0'
gold_color = '#FFD700'
text_color = '#2C3E50'
arrow_color = '#34495E'

# Title
ax.text(9, 11.5, 'Modelo de Dados - Projeto Yelp', 
        fontsize=24, weight='bold', ha='center', color=text_color)
ax.text(9, 11, 'Arquitetura Medalhão (Bronze → Silver → Gold)', 
        fontsize=14, ha='center', color='#7F8C8D', style='italic')

# Helper function to draw table box
def draw_table(ax, x, y, width, height, title, columns, color, layer_label=None):
    # Main box
    box = FancyBboxPatch((x, y), width, height, 
                         boxstyle="round,pad=0.1", 
                         facecolor=color, 
                         edgecolor=text_color, 
                         linewidth=2, 
                         alpha=0.3)
    ax.add_patch(box)
    
    # Title box
    title_box = FancyBboxPatch((x, y + height - 0.6), width, 0.6,
                               boxstyle="round,pad=0.05",
                               facecolor=color,
                               edgecolor=text_color,
                               linewidth=2,
                               alpha=0.8)
    ax.add_patch(title_box)
    
    # Title text
    ax.text(x + width/2, y + height - 0.3, title, 
            fontsize=10, weight='bold', ha='center', va='center', color='white')
    
    # Layer label
    if layer_label:
        ax.text(x + width/2, y + height + 0.3, layer_label, 
                fontsize=9, ha='center', weight='bold', 
                color=color, style='italic',
                bbox=dict(boxstyle='round,pad=0.3', facecolor='white', 
                         edgecolor=color, linewidth=1.5))
    
    # Columns
    y_offset = y + height - 1.1
    for i, col in enumerate(columns[:6]):
        is_pk = col.startswith('🔑')
        is_fk = col.startswith('🔗')
        
        if is_pk or is_fk:
            col_text = col[2:]
            prefix = '🔑 ' if is_pk else '🔗 '
        else:
            col_text = col
            prefix = '• '
        
        font_weight = 'bold' if (is_pk or is_fk) else 'normal'
        ax.text(x + 0.2, y_offset - i*0.35, f'{prefix}{col_text}', 
                fontsize=8, va='top', color=text_color, weight=font_weight)
    
    if len(columns) > 6:
        ax.text(x + 0.2, y_offset - 6*0.35, f'... (+{len(columns)-6} campos)', 
                fontsize=7, va='top', color='#95A5A6', style='italic')

# Helper function to draw arrow
def draw_arrow(ax, x1, y1, x2, y2, label='', style='->'):
    arrow = FancyArrowPatch((x1, y1), (x2, y2),
                           arrowstyle=style, 
                           color=arrow_color,
                           linewidth=2,
                           alpha=0.7,
                           connectionstyle="arc3,rad=0.2")
    ax.add_patch(arrow)
    
    if label:
        mid_x, mid_y = (x1 + x2) / 2, (y1 + y2) / 2
        ax.text(mid_x, mid_y, label, 
                fontsize=8, ha='center', 
                bbox=dict(boxstyle='round,pad=0.3', facecolor='white', 
                         edgecolor=arrow_color, linewidth=1))

# === BRONZE LAYER ===
bronze_y = 7
for i, table in enumerate(input_config['bronze']['tables']):
    x_pos = 0.5 + i * 4.5
    draw_table(ax, x_pos, bronze_y, 3.5, 3,
              table['name'],
              table['columns'],
              bronze_color, 'BRONZE')

# === SILVER LAYER ===
silver_y = 3.5
for i, table in enumerate(input_config['silver']['tables']):
    x_pos = 2 + i * 4.5
    draw_table(ax, x_pos, silver_y, 3.5, 3,
              table['name'],
              table['columns'],
              silver_color, 'SILVER')

# === GOLD LAYER ===
gold_y = 0
for i, table in enumerate(input_config['gold']['tables']):
    x_pos = 0.3 + i * 4.5
    draw_table(ax, x_pos, gold_y, 3.8 if i == 0 else 3.5, 3,
              table['name'],
              table['columns'],
              gold_color, 'GOLD')

# === ARROWS - Bronze to Silver ===
draw_arrow(ax, 2.25, bronze_y, 3.75, silver_y + 3, 'Enriquecimento\ne Validação')

# === ARROWS - Silver to Gold ===
draw_arrow(ax, 3.75, silver_y, 2.2, gold_y + 3, 'Agregação\ne Análise')
draw_arrow(ax, 3.75, silver_y, 6.25, gold_y + 3, 'Métricas de\nMarketing')

# === LEGEND ===
legend_y = 10.2
ax.text(0.5, legend_y, '🔑 Chave Primária    🔗 Chave Estrangeira    • Campo Regular', 
        fontsize=10, color=text_color,
        bbox=dict(boxstyle='round,pad=0.5', facecolor='#ECF0F1', 
                 edgecolor=text_color, linewidth=1.5))

# === METADATA ===
info_text = f"""Schemas: 
• {input_config['bronze']['schema']}
• {input_config['silver']['schema']}
• {input_config['gold']['schema']}

Camadas:
• BRONZE: Dados brutos ingeridos
• SILVER: Dados limpos e enriquecidos  
• GOLD: Dados prontos para negócio"""

ax.text(13.5, 9.5, info_text, 
        fontsize=9, color=text_color, va='top',
        bbox=dict(boxstyle='round,pad=0.7', facecolor='#F8F9FA', 
                 edgecolor=text_color, linewidth=1.5, alpha=0.9),
        family='monospace')

# Add data flow indicators
ax.text(14.5, 6.5, 'Fluxo de Dados', fontsize=11, weight='bold', 
        ha='center', color=text_color)
for i, (layer, desc) in enumerate([
    ('Bronze', 'Ingestão Raw'),
    ('↓', ''),
    ('Silver', 'Limpeza'),
    ('↓', ''),
    ('Gold', 'Agregação')
]):
    y_pos = 6 - i*0.6
    if layer in ['↓']:
        ax.text(14.5, y_pos, layer, fontsize=16, ha='center', color=arrow_color)
    else:
        color = {'Bronze': bronze_color, 'Silver': silver_color, 'Gold': gold_color}[layer]
        ax.text(14.5, y_pos, layer, fontsize=10, ha='center', weight='bold',
               bbox=dict(boxstyle='round,pad=0.4', facecolor=color, 
                        edgecolor=text_color, linewidth=1.5, alpha=0.6))
        if desc:
            ax.text(14.5, y_pos - 0.25, desc, fontsize=8, ha='center', 
                   color='#7F8C8D', style='italic')

plt.tight_layout()
display(fig)

print("✅ Diagrama do modelo de dados gerado com sucesso!")
print(f"\n📊 Schemas configurados:")
print(f"   • Bronze: {input_config['bronze']['schema']}")
print(f"   • Silver: {input_config['silver']['schema']}")
print(f"   • Gold: {input_config['gold']['schema']}")
```
