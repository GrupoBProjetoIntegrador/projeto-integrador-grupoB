#review por localidade#

```python
from pyspark.sql.functions import count

silver_business = spark.table("workspace.yelp_ing.silver_business")
silver_review = spark.table("workspace.yelp_ing.silver_review")

df = (
    silver_business
    .join(silver_review, silver_business.business_id == silver_review.business_id, "inner")
    .groupBy(silver_business.state, silver_business.city)
    .agg(count(silver_review.review_id).alias("quantidade_review"))
    .select(
        silver_business.state.alias("estado"),
        silver_business.city.alias("cidade"),
        "quantidade_review"
    )
)

display(df)
```
