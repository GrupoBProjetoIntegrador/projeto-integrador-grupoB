```python
import os
import requests
import io
import zipfile
import pandas as pd

url = 'https://www.kaggle.com/api/v1/datasets/download/yelp-dataset/yelp-dataset/'
zip_content = None

print(f"Attempting to download from URL: {url}")
response = requests.get(url)

if response.status_code == 200:
    zip_content = io.BytesIO(response.content)
    print("Successfully downloaded ZIP file from URL.")
else:
    print(f"Failed to download from URL. Status code: {response.status_code}. Checking for local archive.zip...")
    local_zip_path = '/content/archive.zip'
    if os.path.exists(local_zip_path):
        with open(local_zip_path, 'rb') as f:
            zip_content = io.BytesIO(f.read())
        print(f"Successfully loaded local ZIP file from {local_zip_path}.")
    else:
        print(f"Local file {local_zip_path} not found. Please upload 'archive.zip' to /content/.")

json_data = None
```

```python
if zip_content:
    try:
        zip_file = zipfile.ZipFile(zip_content)
        for file in zip_file.namelist():
            if file == 'yelp_academic_dataset_business.json':
                with zip_file.open(file) as json_file:
                    json_data = json_file.read().decode('utf-8')
                print(f'Successfully extracted and read {file}. Data size: {len(json_data)} bytes')
                break
        if json_data is None:
            print('yelp_academic_dataset_business.json not found in the zip file.')
    except zipfile.BadZipFile:
        print("Error: The downloaded/local file is not a valid ZIP archive.")
else:
    print("No ZIP content available to process.")
```

```python
target_path = "/Volumes/workspace/yelp_ing/landing_zone_kaggle_api/yelp_academic_dataset_business.json"

# Garantindo que a pasta existe
os.makedirs(os.path.dirname(target_path), exist_ok=True)

# Escrevendo a string da memória para o disco
with open(target_path, "w", encoding="utf-8") as f:
    f.write(json_data)

print(f"Arquivo salvo fisicamente em: {target_path}")
```
