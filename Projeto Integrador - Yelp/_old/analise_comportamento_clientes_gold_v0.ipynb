#analise do comportamento de clientes#

```python
query = """
SELECT
    u.name AS user_name,
    u.user_id,
    b.food_category,
    MAX(r.stars) AS max_stars,
    AVG(r.stars) AS avg_stars,
    MIN(r.stars) AS min_stars,
    COUNT(r.review_id) AS review_count
FROM workspace.yelp_ing.silver_user u
JOIN workspace.yelp_ing.silver_review r ON u.user_id = r.user_id
JOIN workspace.yelp_ing.silver_business b ON r.business_id = b.business_id
GROUP BY u.name, u.user_id, b.food_category
"""

df = spark.sql(query)
display(df)
```
